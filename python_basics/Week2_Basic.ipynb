```python
# 11일차 (2026.1.31) (json) (인벤 저장, 상점)

import random
import os
import json

from google.colab import drive  # [Colab 전용] 구글 드라이브 연결 코드
drive.mount('/content/drive')
SAVE_PATH = "/content/drive/MyDrive/savefile_v2.json"

monsters = [('코볼트',1), ('킬로펭귄',3), ('루인드플랜트',5), ('골든고블린',7),('아이스뱃',9),('워터풀',13),('피닉스',18),('마왕',25),('마신',35)]   #몬스터 리스트

shop_items = {"포션":50, "텔레포트 스크롤":300}

def monster_damage(monster_name, damage):
  real_dmg = random.randint(damage -2, damage +2)
  return real_dmg


def save_game(floor, hp, name, gold, inventory):                   #저장
  save_data = {
      "floor" : floor,
      "hp" : hp,
      "name" : name,
      "gold" : gold,
      "inventory" : inventory
  }

  with open(SAVE_PATH,'w',encoding='utf-8') as f:
    #dump = 데이터를 json 형식으로 변환해서 파일에 쓰기
    json.dump(save_data, f, ensure_ascii = False, indent = 4)
  print("\n[시스템] 게임 데이터가 저장됐습니다. (savefile_v2.json)")

def load_game():                                 #불러오기
  if os.path.exists(SAVE_PATH):
    with open(SAVE_PATH,'r',encoding='utf-8') as f:
      try:
        save_data = json.load(f)    #파일 내용을 딕셔너리로 복구
        name = save_data["name"]
        print(f"\n[시스템]게임의 정보를 불러왔습니다. 환영합니다, {name}님\n")
        return save_data
      except Exception as e: # [수정] 에러 내용을 e라는 변수에 담음
        print(f"⚠️ 저장된 파일을 읽는 중 오류 발생: {e}") # 에러 메시지 출력
        return None
  else:
    return None

print("\n마왕의 탑 - 비료가 될 것인가 전설이 될 것인가\n") # 제목


saved_data = load_game()

if saved_data:
  floor = saved_data["floor"]
  hp_bar = saved_data["hp"]
  user_name = saved_data["name"]
  gold = saved_data["gold"]    #골드 불러오기
  inventory = saved_data["inventory"]  #인벤토리 불러오기
else:
  print("마왕의 탑, 그곳의 꼭대기에는 전설의 검이 잠들어 있는다는 소문이 돈다...")
  print("수많은 사람들이 아성에 도전했지만 모두 실패하여 마왕 밭의 밑거름이 되어버리고 말았다...")
  print('여기, 또 다른 도전자가 마왕의 탑을 도전하려 한다...')
  user_name = input('당신의 이름은?:')
  floor =1
  hp_bar = 100
  gold = 0
  inventory = {"황금사과": 1}
  print(f'{user_name}은(는) 마왕의 탑에 이르렀다.')





#루프  (오류 잡기)
while floor < 10:                                #10층에 도달하기 전까지
  current_monster = monsters[floor-1]
  monster_name = current_monster[0]
  damage = current_monster[1]
  real_dmg = monster_damage(monster_name,damage)
  print(f"{user_name}은(는) {floor}층에 도달했다!")
  print(f"모험자는 {monster_name}을(를) 만났다! 어떻게 할까?")
  print(f"1번: 싸운다    2번: 다음 층으로 도망친다    3번: 인벤토리를 연다    4번: 긴급상점     5번: 저장한다")

  try:
    choice = int(input("선택지 입력(숫자): "))        #선택지 입력
  except ValueError:
    print("잘못된 입력입니다. 숫자를 입력하십시오\n")    #문자를 입력했을 경우 출력
    continue

  if choice == 1:                     #싸움 선택지
    hp_bar -= real_dmg
    print(f"싸우면서 {real_dmg} 데미지를 받았다! 현재 체력:{hp_bar}")
    if hp_bar > 0:
      win_gold = random.randint(30,100)
      gold += win_gold
      print(f"다음 층으로 올라간다...")
      floor += 1
    else:
      hp_bar = 0
      print(f"{user_name}은(는) 사망했다!")
      break

  elif choice == 2:                     #도망 선택지
    run_percent = random.randint(1,10)
    if run_percent > 4:                #60퍼센트 확률로 실패
        hp_bar -= 10
        print(f"도망에 실패했다! 10데미지를 입었다! 현재 체력:{hp_bar}")
        if hp_bar <= 0:
          hp_bar = 0
          print(f"{user_name}은(는) 사망했다!")
          break
    else:
      print(f"{monster_name}에게서 도망치는데 성공했다!")
      print(f"다음 층으로 올라간다...")
      floor+=1

  elif choice == 3:                    #hp 회복 선택지
    print(f"인벤토리를 열었다! 현재 아이템: {inventory}")
    item_select = input("어느 아이템을 사용할까?(취소 시 원래 화면): ")
    if item_select in inventory:
      if inventory[item_select] > 0:
        if item_select == "포션":
          hp_bar += 20
          if hp_bar >= 100:
            hp_bar = 100
          print(f"{item_select}을(를) 소비하여 hp를 회복했다! 현재 체력: {hp_bar}")

        elif item_select == "황금사과":
          hp_bar += 50
          if hp_bar >= 100:
            hp_bar = 100
          print(f"가문 대대로 내려온 전설의 {item_select}을(를) 소비하여 hp를 회복했다! 현재 체력: {hp_bar}")

        inventory[item_select] -= 1

      else:
        print("남은 아이템 수량이 없습니다!")
    elif item_select == "취소":
      print("취소했습니다.")
    else:
      print("해당 아이템은 없습니다.")

  elif choice == 4:                        #상점 선택지
    print(f"시간을 멈추는 상인이 나타났다! 현재 골드: {gold}")
    buy_name = input("어떤 물건을 살까? 포션: 50, 텔레포트 스크롤: 300(취소 시 원래 화면): ")
    if buy_name == "취소":
      print("취소했습니다.")

    elif gold >= price:
      price = shop_items[buy_name]
      gold -= price                    #물건 사기
      if buy_name in inventory:
        inventory[buy_name] += 1         #기존 아이템일 경우 개수 추가
      else:
        inventory[buy_name] = 1        #새 아이템일 경우 목록에 추가
      print(f"{buy_name}구매에 성공했다!")
    elif gold < price:
      print("골드가 부족합니다!")
    else:
      print("해당 아이템은 팔지 않습니다.")

  elif choice == 5:                  #저장 선택지
    save_game(floor, hp_bar, user_name, gold, inventory)
    print("게임의 내용이 저장되었습니다.")

  else:
    print("잘못된 숫자입니다. 1, 2, 3, 4 중 하나의 숫자를 입력하십시오\n")   #숫자를 잘못 입력했을 경우 출력
    continue

if hp_bar > 0:
  print(f"\n{user_name}은(는) 10층에 도달해서 전설을 검을 구했다!")
  print("게임 클리어!")
else:
  print(f'\n{user_name}은(는) 도전은 끝이 났다... GAME OVER')
```
