```python
# 4일차 (2026.1.22)(6일치 미리 학습, while, for)

# While
import time

monster_hp = 100
damage = 25
turn = 0

print(f"battle start! 오크의 hp: {monster_hp}")

while monster_hp > 0:    #오크의 hp가 0이 될 때까지 반복
  turn += 1
  monster_hp -= damage
  print(f"{turn}번째 공격! 현재 오크의 hp: {monster_hp}")
  time.sleep(0.5)      # 턴마다 .5초 지연

print(f"오크가 쓰러졌다!")

# For
orc_item = ["가죽", "기름", "이빨", "곤봉", "금화"]
print("\n주인공은 다음의 전리품을 얻었다!")
for item in orc_item:
  print(item)

#인벤토리 칸 제한

inventory = []
while len(inventory) < 3 and len(orc_item) > 0:
  item = orc_item.pop(0)
  inventory.append(item)
  print(f"주인공은 {item}을 가방에 넣었다")

print("가방이 가득 찼다!")
print(f"주인공은 {orc_item}을 두고 갔다...\n")


#
inventory = []
orc_item = ["가죽", "기름", "이빨", "곤봉", "금화"]

while True:
  if len(inventory) >= 3:
    print("가방이 가득 찼다!")
    print(f"주인공은 {orc_item}을 두고 갔다...")
    break
  else:
    for item in orc_item:
      item = orc_item.pop(0)
      inventory.append(item)
      print(f"주인공은 {item}을 가방에 넣었다")
```

```python
# 5일차 (2026.1.23) (중간점검)

floor =1


covolt = {"name": "코볼트", "damage": 1}
killer_penguin = {"name": "킬러펭귄", "damage": 3}
ruined_plant = {"name": "루인드플랜트", "damage": 5}
golden_goblin = {"name": "골든고블린", "damage": 9}
ice_bat = {"name": "아이스뱃", "damage": 11}
water_pool = {"name": "워터풀", "damage": 15}
phoenix = {"name": "피닉스", "damage": 20}
demon_king = {"name": "마왕", "damage": 30}
demon_god = {"name": "마신", "damage": 50}
hp_bar = 100

monsters = [covolt, killer_penguin, ruined_plant, golden_goblin,ice_bat,water_pool,phoenix,demon_king,demon_god]

while floor < 10:  #10층에 도달하기 전까지
  print(f"모험자는 {floor}층에 도달했다!")
  print(f"모험자는 {monsters[floor-1]["name"]}을(를) 만났다!")
  choice = input("싸운다 / 도망간다:")
  if choice == "싸운다":
    hp_bar -= monsters[floor-1]['damage']
    if hp_bar > 0:
      print(f"싸우면서 {monsters[floor-1]['damage']} 데미지를 받았다! 현재 체력: {hp_bar}")
      floor += 1
    else:
      hp_bar = 0
      print(f"싸우면서 {monsters[floor-1]['damage']} 데미지를 받았다! 현재 체력: {hp_bar}")
      print("모험자는 사망했다!")
      break

  elif choice == "도망간다":
    hp_bar -= 10
    print(f"도망치면서 10의 데미지를 받았다! 현재 체력: {hp_bar}")
    if hp_bar > 0:
      print(f"도망치면서 10의 데미지를 받았다! 현재 체력: {hp_bar}")
      floor += 1
    else:
      hp_bar = 0
      print(f"도망치면서 10의 데미지를 받았다! 현재 체력: {hp_bar}")
      print("모험자는 사망했다!")
      break
  else:
    print("오류! 다시 적으십시오.")

if hp_bar > 0:
  print("\n모험자는 10층에 도달해서 전설을 검을 구했다!")
else:
  print('\n모험자의 모험은 끝이 났다... GAME OVER')
```

```python
# 7일차 (2026.1.25) (definition) (기존 코드 정리)

floor =1
hp_bar = 100

monsters = [('코볼트',1), ('킬로펭귄',3), ('루인드플랜트',5), ('골든고블린',9),('아이스뱃',11),('우터풀',15),('피닉스',20),('마왕',30),('마신',50)]   #몬스터 리스트



def monster_battle(monster_name, damage):       #definition, 배틀에 대한 정의
  global hp_bar
  hp_bar -= damage
  print(f"모험자는 {monster_name}을(를) 만났다!")
  print(f"싸우면서 {damage} 데미지를 받았다! 현재 체력:{hp_bar}")



while floor < 10:                                #10층에 도달하기 전까지
  print(f"모험자는 {floor}층에 도달했다!")
  current_monster = monsters[floor-1]            #현재 층수에 있는 몬스터
  monster_name = current_monster[0]              #몬스터 이름
  damage = current_monster[1]                    #몬스터 데미지
  monster_battle(monster_name,damage)
  if hp_bar > 0:
      print(f"다음 층으로 올라간다")
      floor += 1
  else:
      hp_bar = 0
      print("모험자는 사망했다!")
      break

if hp_bar > 0:
  print("\n모험자는 10층에 도달해서 전설을 검을 구했다!")
else:
  print('\n모험자의 모험은 끝이 났다... GAME OVER')
```

```python
# 8일차 (2026.1.26) (random, input) (기존 코드 정리)
import random

floor =1
hp_bar = 100

monsters = [('코볼트',1), ('킬로펭귄',3), ('루인드플랜트',5), ('골든고블린',7),('아이스뱃',9),('워터풀',13),('피닉스',18),('마왕',25),('마신',35)]   #몬스터 리스트

def monster_damage(monster_name, damage):       #definition, 배틀에 대한 정의
  print(f"모험자는 {monster_name}을(를) 만났다!")
  print(f"싸우면서 {damage} 데미지를 받았다! 현재 체력:{hp_bar}")
  real_dmg = random.randit(damage +- 2)
  return real_dmg

print("마왕의 탑, 그곳의 꼭대기에는 전설의 검이 잠들어 있는다는 소문이 돈다...")
print("수많은 사람들이 아성에 도전했지만 모두 실패하여 마왕 밭의 밑거름이 되어버리고 말았다...")
print('여기, 또 다른 도전자가 마왕의 탑을 도전하려 한다...')
user_name = input('당신의 이름은?:')
print(f'{user_name}은(는) 마왕의 탑에 이르렀다.')


while floor < 10:                                #10층에 도달하기 전까지
  print(f"{user_name}은(는) {floor}층에 도달했다!")
  current_monster = monsters[floor-1]            #현재 층수에 있는 몬스터
  monster_name = current_monster[0]              #몬스터 이름
  damage = current_monster[1]                    #몬스터 데미지
  monster_battle(monster_name,damage)

  if hp_bar > 0:
    print(f"다음 층으로 올라간다...")
    floor += 1
  else:
      hp_bar = 0
      print(f"{user_name}은(는) 사망했다!")
      break

if hp_bar > 0:
  print(f"\n{user_name}은(는) 10층에 도달해서 전설을 검을 구했다!")
else:
  print(f'\n{user_name}은(는) 도전은 끝이 났다... GAME OVER')
```
